```python
csv_path = files_found[0]

fertilizer_df = pd.read_csv(csv_path)

print("Shape:", fertilizer_df.shape)
print("\nColumns:")
print(fertilizer_df.columns.tolist())

print("\nFirst 5 rows:")
display(fertilizer_df.head())

print("\nData types:")
print(fertilizer_df.dtypes)
```

```python
# Step 4: Inspect complete fertilizer dataset

pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", None)

print("Dataset Shape:")
print(fertilizer_df.shape)

print("\n==============================")
print("ALL 20 COLUMNS")
print("==============================")

for i, col in enumerate(fertilizer_df.columns):
    print(f"{i:02d} → {col}")

print("\n==============================")
print("DATA TYPES")
print("==============================")

print(fertilizer_df.dtypes)

print("\n==============================")
print("MISSING VALUES")
print("==============================")

print(fertilizer_df.isnull().sum())

print("\n==============================")
print("FIRST 5 ROWS")
print("==============================")

display(fertilizer_df.head())
```

```python
# Step 5: Unique values of categorical columns

categorical_columns = fertilizer_df.select_dtypes(
    include=["object", "category"]
).columns

print("Categorical columns:")
print(list(categorical_columns))

for col in categorical_columns:
    print("\n--------------------------------")
    print(col)
    print("--------------------------------")
    print("Unique values:", fertilizer_df[col].nunique())
    print(fertilizer_df[col].unique()[:30])
```

```python
# Step 7: Recommended fertilizer distribution

target_column = "Recommended_Fertilizer"

print("Target distribution:")
print(
    fertilizer_df[target_column].value_counts()
)

print("\nTarget percentages:")
print(
    fertilizer_df[target_column]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

fertilizer_df[target_column].value_counts().plot(
    kind="bar"
)

plt.xlabel("Fertilizer")
plt.ylabel("Number of Samples")
plt.title("Recommended Fertilizer Distribution")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
# Step 8: Data quality check

print("Missing values:")
print(fertilizer_df.isnull().sum())

print("\nTotal missing values:")
print(fertilizer_df.isnull().sum().sum())

print("\nDuplicate rows:")
print(fertilizer_df.duplicated().sum())
```

```python
# Step 9: Separate X and y

X = fertilizer_df.drop(
    columns=["Recommended_Fertilizer"]
)

y = fertilizer_df["Recommended_Fertilizer"]

print("Feature shape:", X.shape)
print("Target shape:", y.shape)

print("\nFeatures:")
print(X.columns.tolist())

print("\nTarget classes:")
print(sorted(y.unique()))
```

```python
# Step 10: Feature groups

categorical_features = X.select_dtypes(
    include=["object", "category"]
).columns.tolist()

numerical_features = X.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

print("Categorical features:")
print(categorical_features)

print("\nNumerical features:")
print(numerical_features)
```

```python
# Step 11: Train-test split

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))

print("\nTraining distribution:")
print(y_train.value_counts())

print("\nTesting distribution:")
print(y_test.value_counts())
```

```python
# Step 12: Create preprocessing pipeline

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore"
            ),
            categorical_features
        ),
        (
            "numerical",
            "passthrough",
            numerical_features
        )
    ]
)

print("Preprocessor created successfully.")
```

```python
# Step 13: Create Random Forest model

from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline

rf_model = RandomForestClassifier(
    n_estimators=300,
    max_depth=None,
    min_samples_split=2,
    min_samples_leaf=1,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

fertilizer_pipeline = Pipeline(
    steps=[
        (
            "preprocessor",
            preprocessor
        ),
        (
            "classifier",
            rf_model
        )
    ]
)

print("Random Forest pipeline created.")
```

```python
# Step 14: Train fertilizer recommendation model

print("Training Fertilizer AI...")
print("Please wait...")

fertilizer_pipeline.fit(
    X_train,
    y_train
)

print("\n✅ Training completed!")
```

```python
# Step 15: Evaluate model

train_accuracy = fertilizer_pipeline.score(
    X_train,
    y_train
)

test_accuracy = fertilizer_pipeline.score(
    X_test,
    y_test
)

print("==============================")
print("FERTILIZER MODEL RESULTS")
print("==============================")

print(f"Training Accuracy: {train_accuracy * 100:.2f}%")
print(f"Testing Accuracy:  {test_accuracy * 100:.2f}%")
```

```python
# Step 16: Classification report

from sklearn.metrics import classification_report

y_pred = fertilizer_pipeline.predict(X_test)

print(
    classification_report(
        y_test,
        y_pred,
        digits=4
    )
)
```

```python
# Step 17: Confusion matrix

from sklearn.metrics import confusion_matrix
import seaborn as sns

cm = confusion_matrix(
    y_test,
    y_pred,
    labels=sorted(y.unique())
)

plt.figure(figsize=(9, 7))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Greens",
    xticklabels=sorted(y.unique()),
    yticklabels=sorted(y.unique())
)

plt.xlabel("Predicted Fertilizer")
plt.ylabel("Actual Fertilizer")
plt.title("CROPORA Fertilizer Recommendation - Confusion Matrix")

plt.tight_layout()
plt.show()
```

```python
# Step 18: Organic / Inorganic fertilizer mapping

ORGANIC_FERTILIZERS = {
    "Compost"
}

INORGANIC_FERTILIZERS = {
    "MOP",
    "Urea",
    "Zinc Sulphate",
    "NPK",
    "DAP",
    "SSP"
}

def get_fertilizer_type(fertilizer_name):

    if fertilizer_name in ORGANIC_FERTILIZERS:
        return "Organic"

    elif fertilizer_name in INORGANIC_FERTILIZERS:
        return "Inorganic"

    else:
        return "Unknown"


print("Compost:", get_fertilizer_type("Compost"))
print("Urea:", get_fertilizer_type("Urea"))
print("NPK:", get_fertilizer_type("NPK"))
```

```python
# Step 19: Test one sample

sample = X_test.iloc[[0]]

actual = y_test.iloc[0]

prediction = fertilizer_pipeline.predict(
    sample
)[0]

probabilities = fertilizer_pipeline.predict_proba(
    sample
)[0]

print("==============================")
print("FERTILIZER PREDICTION")
print("==============================")

print("Actual:", actual)
print("Predicted:", prediction)

print(
    "Type:",
    get_fertilizer_type(prediction)
)

print(
    "Confidence:",
    f"{max(probabilities) * 100:.2f}%"
)
```

```python
# Step 20: Top 3 fertilizer recommendations

def predict_top_fertilizers(sample):

    probabilities = fertilizer_pipeline.predict_proba(sample)[0]

    classes = fertilizer_pipeline.classes_

    top_indices = np.argsort(probabilities)[::-1][:3]

    recommendations = []

    for index in top_indices:

        fertilizer = classes[index]
        confidence = probabilities[index]

        recommendations.append({
            "fertilizer": fertilizer,
            "type": get_fertilizer_type(fertilizer),
            "confidence": round(confidence * 100, 2)
        })

    return recommendations


recommendations = predict_top_fertilizers(
    X_test.iloc[[0]]
)

for item in recommendations:
    print(item)
```
